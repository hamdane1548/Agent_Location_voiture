```python
import sys
from pathlib import Path

# Racine de agent_backend
project_root = Path.cwd().parent
print(project_root)
sys.path.insert(0, str(project_root))

from Settings import Settings

settings = Settings()

print("Loaded:", settings.MISTRAL_AI_API_key)
```

```python
# check file send into service of 
project_root = Path.cwd().parent  / "Embeddings"
from loguru import logger
print(project_root)
sys.path.insert(0, str(project_root))
from Embedding import RAG_Embedding
#test send file into Rag_Embeddings_function
pdf_path = Path(
    r"C:\Users\oussa\Documents\ai_project\ai_project\Location_voiture\agent_backend\src\Data\Documentation Opoo Car Rental.pdf"
)
rag = RAG_Embedding(file_path=pdf_path,file=None)
logger.info(f"the number of pages into the pdf {len(rag.pages)}")
#content_chunking : str = ""
content_chunking = RAG_Embedding.chunkingDocuments(fiels=rag)
spliting = content_chunking.replace('\n',' ').split('. ')
logger.info(f"the value of the content is {type(spliting)}")

chunks = RAG_Embedding.spliting_content(spliting)
logger.info(f"the value of the chunks is {len(chunks)}")
```

```python
project_root = Path.cwd().parent  / "Model"
sys.path.insert(0, str(project_root))


chunks_list= []
for i in chunks:
    vector =  RAG_Embedding.embedding_using_transfromes_model_encoding(i)
    logger.info(f"the cotentis is {i} and the vector is {vector}")
    chunks_list.append([vector,i])
        
logger.info(f"the content is {chunks_list}")
```

## Save the embedding ( vector ) in the vector data base store

```python
# call the vectore store save function
for x,y in chunks_list:
    logger.info(f"the value of the content is {x}")
    logger.info(f"the content is {y}")
    RAG_Embedding.SaveInVectorDb(vector=x,content=y,collection_name="rag-location",id_client=1,copmany_name="location_voiture",name_cleint="infinity")
```
